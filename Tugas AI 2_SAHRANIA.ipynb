# 5. OUTPUT PROMPT 5 — PENGOLAHAN DATA GEOFISIKA MENGGUNAKAN PYTHON

### 5.1. Identifikasi File dan Struktur Data
* Pemrosesan sinyal seismik 1D digital sintetis dengan kontaminasi derau lingkungan (*Gaussian noise*).
* Format: Deret waktu ($t$) berfrekuensi cuplik $1000\ \text{Hz}$.

### 5.2. Penjelasan Tujuan Analisis
1. Mensintesis respon Ricker Wavelet $30\ \text{Hz}$.
2. Mengaplikasikan penapisan pita frekuensi (*Butterworth Bandpass Filter* $10 - 50\ \text{Hz}$).
3. Menghitung peningkatan *Signal-to-Noise Ratio* (SNR).
4. Menampilkan visualisasi komparatif sebelum dan sesudah filtrasi.

### 5.3. Penjelasan Metode yang Digunakan
* **Ricker Wavelet:** $f(t) = (1 - 2\pi^2 f_0^2 t^2) e^{-\pi^2 f_0^2 t^2}$
* **Butterworth Zero-Phase Bandpass (`filtfilt`):** Mencegah distorsi pergeseran fasa sinyal seismik.

### 5.4. Kode Python Lengkap

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy import signal


# 1. PARAMETER WAKTU DAN PEMBUATAN SINYAL
def buat_data_seismik(durasi=0.5, fs=1000, f0=30):
    t = np.linspace(-durasi / 2, durasi / 2, int(durasi * fs), endpoint=False)
    arg = np.pi**2 * f0**2 * t**2
    wavelet = (1.0 - 2.0 * arg) * np.exp(-arg)

    np.random.seed(42)
    noise = np.random.normal(0, 0.25, len(t))
    sinyal_terkontaminasi = wavelet + noise
    return t, wavelet, sinyal_terkontaminasi, fs


# 2. FILTER DIGITAL BUTTERWORTH
def butter_bandpass_filter(data, lowcut, highcut, fs, order=4):
    nyquist = 0.5 * fs
    low = lowcut / nyquist
    high = highcut / nyquist
    b, a = signal.butter(order, [low, high], btype='band')
    y = signal.filtfilt(b, a, data)
    return y


# 3. PROSES DATA
t, wavelet, sinyal_noise, fs = buat_data_seismik()
sinyal_bersih = butter_bandpass_filter(
    sinyal_noise, lowcut=10.0, highcut=50.0, fs=fs
)

# 4. VISUALISASI
plt.figure(figsize=(10, 7))

plt.subplot(3, 1, 1)
plt.plot(t, wavelet, color='blue', label='Sinyal Asli (Ricker 30 Hz)')
plt.ylabel('Amplitudo')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()

plt.subplot(3, 1, 2)
plt.plot(t, sinyal_noise, color='red', alpha=0.7, label='Sinyal + Noise')
plt.ylabel('Amplitudo')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()

plt.subplot(3, 1, 3)
plt.plot(
    t,
    sinyal_bersih,
    color='green',
    linewidth=1.8,
    label='Sinyal Hasil Filter (10-50 Hz)',
)
plt.xlabel('Waktu (detik)')
plt.ylabel('Amplitudo')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()

plt.tight_layout()
plt.show()

# 5. METRIK KUALITAS
snr_awal = 10 * np.log10(np.sum(wavelet**2) / np.sum((sinyal_noise - wavelet) ** 2))
snr_akhir = 10 * np.log10(
    np.sum(wavelet**2) / np.sum((sinyal_bersih - wavelet) ** 2)
)
print(f'SNR Sebelum Filter : {snr_awal:.2f} dB')
print(f'SNR Sesudah Filter : {snr_akhir:.2f} dB')
```

### 5.5. Penjelasan Kode per Bagian
* `buat_data_seismik()`: Membangun sinyal murni dan menambahkan noise acak Gaussian.
* `signal.butter` & `signal.filtfilt`: Merancang koefisien filter IIR Butterworth dan menerapkannya secara bolak-balik tanpa distorsi fasa.
* Evaluasi SNR: Menghitung daya sinyal relatif terhadap residu derau.
